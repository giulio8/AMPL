```python
import matplotlib.pyplot as plt

# Dati
scenarios = [1, 2, 3, 4, 5]
UD = [27.5, 27.5, 27.5, 27.5, 27.5]
UA = [6.89, 6.89, 6.89, 6.89, 6.89]

# Primo grafico: Utility del Defender
plt.figure()
plt.plot(scenarios, UD, marker='o')
plt.xlabel('Scenario')
plt.ylabel('Defender Utility (UD)')
plt.title('Defender Utility across Scenarios')
plt.grid(True)
plt.xticks(scenarios)
plt.show()

# Secondo grafico: Utility dell'Attacker
plt.figure()
plt.plot(scenarios, UA, marker='s')
plt.xlabel('Scenario')
plt.ylabel('Attacker Utility (UA)')
plt.title('Attacker Utility across Scenarios')
plt.grid(True)
plt.xticks(scenarios)
plt.show()
```
